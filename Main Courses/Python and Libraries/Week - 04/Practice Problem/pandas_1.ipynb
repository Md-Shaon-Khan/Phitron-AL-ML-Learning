```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    'A': [1, 2, np.nan, 4, 5],
    'B': [5, np.nan, np.nan, None, 10],
    'C': ['a', 'b', 'c', None, 'e']
})
                   
print(df)  
```

```python
df_filled = df.fillna({
    'A':0,
    'B':99,
    'C':'Unknown'
})

df_filled
```

```python
df_fill_B = df.copy()

df_fill_B
```

```python
df_fill_B['B'] = df_fill_B['B'].ffill()
df_fill_B
```

```python
def fill_missing(df):
    df_copy = df.copy()

    for column in df_copy.columns:
        if df_copy[column].dtype in ['float64','int64']:
            df_copy[column] = df[column].fillna(df_copy[column].mean())
        else:
            df_copy[column] = df_copy[column].fillna('Unknown')
        
    return df_copy

df_custom = fill_missing(df)
df_custom
```

```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    'A': [1, 2, np.nan, 4, 5],
    'B': [5, np.nan, np.nan, None, 10],
    'C': ['a', 'b', 'c', None, 'e']
})
                   
print(df)
```

```python
# Drop rows where column A has missing values
df_clean_based_A = df.dropna(subset=['A'])
print(df_clean_based_A)



# Drop rows with less than 2 non-null values
df_clean_based_2 = df.dropna(thresh=2)
print(df_clean_based_2)


# Drop rows where both Column A and Column B has missing values
df_clean_based_AB = df.dropna(subset=['A', 'B'], how='all')
print(df_clean_based_AB)
```
