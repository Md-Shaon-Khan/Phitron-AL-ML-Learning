```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
sale = pd.read_csv('sales.csv')
sale
```

```python
sns.set_style('whitegrid')
```

```python
plt.figure(figsize = (30, 10))
sns.lineplot(data = sale, x = 'date', y = 'sales')

plt.title('Sales Over Time')
plt.xticks(rotation = 90)
plt.show()
```

```python
students = pd.read_csv('students.csv')
students
```

```python
sns.scatterplot(
    data = students,
    x = 'hours_studied',
    y = 'marks',
    hue = 'gender'
)
plt.title('Hours Studied vs Marks')
plt.show()
```

```python
g = sns.FacetGrid(students, col='subject')

g.map_dataframe(
    sns.scatterplot,
    x='hours_studied',
    y='marks'
)

g.fig.suptitle("Hours Studied vs Marks by Subject", y=1.05)

plt.show()
```

```python
sns.histplot(
    data = students,
    x = 'marks',
    kde = True
)
```

```python
sns.kdeplot(
    data = students,
    x = 'hours_studied',
    fill = True
)
plt.title('Distribution of Hours Studied')
plt.show()
```

```python
sns.countplot(
    data = students,
    x = 'subject'
)
```

```python
sns.regplot(
    data = students,
    x = 'hours_studied',
    y = 'marks'
)
```

```python
sns.pairplot(
    students,
    hue = 'gender'
)
```
