```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
hours = [2,3,4,1,5,7,8]
days = [1,3,4,5,6,7,8]

plt.plot(days, hours)
```

```python
plt.xlabel('Days')
plt.ylabel('Study Hours')
plt.title('Tread of a student study hours')
plt.plot(days, hours)
```
