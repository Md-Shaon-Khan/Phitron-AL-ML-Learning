```python
nltk.download('punkt')
nltk.download('punkt_tab')
```

```python
tokens = word_tokenize(document.lower())
tokens
```

```python
vocab = {'<unk>': 0}

for token in Counter(tokens).keys():
    if token not in vocab:
        vocab[token] = len(vocab)

vocab
```

```python
input_sentences = document.split('\n')
input_sentences
```

```python
input_sentences[0]
```

```python
def text_to_indices(sentence, vocab):
    numerical_sentence = []

    for token in sentence:
        if token in vocab:
            numerical_sentence.append(vocab[token])
        else:
            numerical_sentence.append(vocab['<unk>'])
    return numerical_sentence
```

```python
input_numerical_sentence = []

for sentence in input_sentences:
    numerical_sentence = text_to_indices(word_tokenize(sentence.lower()), vocab)
    input_numerical_sentence.append(numerical_sentence)
```

```python
hello = "The Last Train Journey"
ok = text_to_indices(hello.lower().split(), vocab)
ok
```

```python
input_numerical_sentence
```

```python
training_sequence = []
for sentence in input_numerical_sentence:
    for i in range(1, len(sentence)):
        training_sequence.append(sentence[:i+1])
```

```python
training_sequence
```

```python
len(training_sequence)
```

```python
len_list = []

for sequence in training_sequence:
    len_list.append(len(sequence))

max(len_list)
```

```python
[0] * (max(len_list) - len(sequence))
```

```python
padding_training_sequence = []

for sequence in training_sequence:
    padding_training_sequence.append([0] * (max(len_list) - len(sequence)) + sequence)
```

```python
len(padding_training_sequence[1])
```

```python
padded_training_sequence = torch.tensor(padding_training_sequence, dtype=torch.long)
padded_training_sequence
```

```python
X = padded_training_sequence[:, :-1]
y = padded_training_sequence[:, -1]
```

```python
len(X), len(y)
```

```python
class CustomDataset(Dataset):
    def __init__(self, X, y):
        self.X = X
        self.y = y

    def __len__(self):
        return self.X.shape[0]

    def __getitem__(self, idx):
        return self.X[idx], self.y[idx]
```

```python
dataset = CustomDataset(X, y)
len(dataset)
```

```python
dataloader = DataLoader(dataset, batch_size=32, shuffle=True)

for input, target in dataloader:
    print("Input:", input)
    print("Target:", target)
    break
```
