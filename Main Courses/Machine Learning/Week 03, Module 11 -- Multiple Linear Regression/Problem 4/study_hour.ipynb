```python
import numpy as np
```

```python
X = np.array([
    [4, 7, 85],
    [5, 7.5, 95],
    [6, 8, 100],
    [3, 10, 75],
    [1, 12, 55],
    [7, 7, 100],
    [8, 6, 95],
    [7.5, 8, 100],
    [12, 9, 90],
    [3, 8, 100]
])

y = np.array([75, 78, 85, 60, 25, 87, 90, 95, 92, 65])
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
sns.scatterplot(x = X[:,0:1].ravel(), y = y)
```

```python
sns.scatterplot(x = X[:,1:2].ravel(), y = y)
```

```python
sns.scatterplot(x = X[:,2:3].ravel(), y = y)
```

```python
def make_prediction(X, w, b):
    return X @ w + b

w = np.random.rand(3)
b = np.random.rand()
y_pred = make_prediction(X, w, b)
y_pred
```

```python
print(y_pred)
print(y)

print("Error: ", y_pred - y)
```

```python
def compute_cost(X, y, w, b):
    m = len(y)
    y_pred = X @ w + b

    cost = (1 / (2*m)) * np.sum((y_pred - y) ** 2)
    return cost

w_fixed = np.array([0.5, 0.8])
b_fixed = 0.5

w0_range = np.linspace(-10, 10, 500)
cost_history = []

for w0 in w0_range:
    w_current = np.array([w0, w_fixed[0], w_fixed[1]])
    current_cost = compute_cost(X, y, w_current, b_fixed)
    cost_history.append(current_cost)

cost_history = np.array(cost_history)

min_index = np.argmin(cost_history)
best_w0 = w0_range[min_index]
min_cost = cost_history[min_index]

print(f"Minimum cost: {min_cost:.2f}, best w0: {best_w0:.2f}")
```

```python
plt.figure(figsize=(10, 6))
plt.plot(w0_range, cost_history)


plt.scatter(best_w0, min_cost, color='red')
plt.show()
```
