```python
import numpy as np
```

```python
X = np.array([
    [4, 7, 85],
    [5, 7.5, 95],
    [6, 8, 100],
    [3, 10, 75],
    [1, 12, 55],
    [7, 7, 100],
    [8, 6, 95],
    [7.5, 8, 100],
    [12, 9, 90],
    [3, 8, 100]
])

y = np.array([75, 78, 85, 60, 25, 87, 90, 95, 92, 65])
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
sns.scatterplot(x = X[:,0:1].ravel(), y = y)
```

```python
def make_prediction(X, w, b):
    return X @ w + b

w = np.random.rand(3)
b = np.random.rand()
y_pred = make_prediction(X, w, b)
y_pred
```

```python
print(y_pred)
print(y)

print("Error: ", y_pred - y)
```
