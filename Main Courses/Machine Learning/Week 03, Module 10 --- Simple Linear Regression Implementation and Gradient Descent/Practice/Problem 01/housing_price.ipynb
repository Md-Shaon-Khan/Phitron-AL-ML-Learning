```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("F:\Jupyter Projects\Phitron AL ML\Dataset\Housing.csv")
df.head()
```

```python
price_col = np.array(df['price'])
area_col = np.array(df['area'])

area_col = (area_col - area_col.min()) / (area_col.max() - area_col.min())
price_col = (price_col - price_col.min()) / (price_col.max() - price_col.min())

display(price_col)
display(area_col)
```

```python
sns.scatterplot( x = area_col, y = price_col)
```

```python
def make_prediction(area, w, b):
    pred_list = area * w + b
    return pred_list
```

```python
predictions = make_prediction(area_col, 0.1, 0.1)
display(predictions)

sns.scatterplot(x = area_col, y = price_col)
plt.plot(area_col, predictions, color = 'red')
```

```python
def compute_cost(predictions, price_col):
    cost = np.mean((predictions - price_col) ** 2)
    return cost
```

```python
w = 0.2
b = 0.3

predictions = make_prediction(area_col, w, b)
cost = compute_cost(predictions, price_col)
print(cost)

sns.scatterplot(x = area_col, y = price_col)
plt.plot(area_col, predictions, color = 'red')
plt.title(f"Cost = {cost} when w = {w} and b = {b}")
```

```python
w = 0.3
b = 0.05

predictions = make_prediction(area_col, w, b)
cost = compute_cost(predictions, price_col)
print(cost)

sns.scatterplot(x = area_col, y = price_col)
plt.plot(area_col, predictions, color = 'red')
plt.title(f"Cost = {cost} when w = {w} and b = {b}")
```

```python
w = []
cost = []

for i in range(-10, 10):
    cost.append(compute_cost(make_prediction(area_col, i, b = 0), price_col))
    w.append(i)

plt.plot(w, cost)
plt.grid()
```

```python
def calculate_gradient(area, price, w, b):
    n = len(area)

    dj_dw = 0.0
    dj_db = 0.0

    for i in range(n):
        pred = make_prediction(area[i], w, b)
        error = pred - price[i]
        dj_dw += (error * area[i])
        dj_db += error
    
    dj_dw = (dj_dw * 2) / n
    dj_db = (dj_db * 2) / n

    return dj_dw, dj_db
```

```python
def gradient_descent(area, price, w_in, b_in, max_iters, alpha = 0.01):
    w = w_in
    b = b_in

    cost_history = []
    for i in range(max_iters):
        dj_dw, dj_db = calculate_gradient(area, price, w, b)
        w = w - alpha * dj_dw
        b = b - alpha * dj_db

        
        if i % 2 == 0:
            current_cost = compute_cost(make_prediction(area, w, b), price)
            cost_history.append(current_cost)
            
            print(f"Iteration {i:3d}: when w = {w:.6f}, b = {b:.6f}, cost = {current_cost:.6f}")
            
    return w, b, cost_history


w, b, cost_history = gradient_descent(area_col, price_col, 0, 0, 1000, alpha = 0.01)
```
