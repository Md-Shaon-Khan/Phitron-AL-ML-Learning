## Question 1: What is the fraud percentage in this dataset?

```python
total_transactions = len(df)
fraud_count = df['Class'].sum()
fraud_percentage = (fraud_count / total_transactions) * 100

print(f"Total transactions: {total_transactions}")
print(f"Fraud transactions: {fraud_count}")
print(f"Fraud percentage: {fraud_percentage:.4f}%")
```

# Question 2: Train K-Means with K=10 and calculate distance to nearest cluster center

```python
# Separate features and labels
X = df.drop('Class', axis=1)
y = df['Class']

# Scale features (K-Means is distance-based, so scaling is essential)
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Train K-Means with K=10
kmeans = KMeans(n_clusters=10, random_state=42, n_init=10)
kmeans.fit(X_scaled)

# Get cluster centers
centers = kmeans.cluster_centers_

# Calculate distance of each point to its nearest cluster center
labels = kmeans.labels_
distances = np.linalg.norm(X_scaled - centers[labels], axis=1)

df['distance_to_center'] = distances
```

# Question 3: Set threshold at 95th percentile — how many frauds are detected?

```python
# Set threshold at 95th percentile of distances
threshold = np.percentile(distances, 95)
print(f"Threshold (95th percentile): {threshold:.4f}")

# Flag points beyond threshold as anomalies
df['predicted_anomaly'] = (df['distance_to_center'] > threshold).astype(int)

# Count how many actual frauds are detected
frauds_detected = df[(df['predicted_anomaly'] == 1) & (df['Class'] == 1)].shape[0]
total_frauds = df[df['Class'] == 1].shape[0]

print(f"Total frauds in dataset: {total_frauds}")
print(f"Frauds detected as anomalies: {frauds_detected}")
```

# Question 4: Precision and Recall for fraud detection

```python
precision = precision_score(df['Class'], df['predicted_anomaly'])
recall = recall_score(df['Class'], df['predicted_anomaly'])

print(f"Precision: {precision:.4f}")
print(f"Recall: {recall:.4f}")

# Optional: full confusion matrix for clarity
cm = confusion_matrix(df['Class'], df['predicted_anomaly'])
print("Confusion Matrix:\n", cm)
```

# Question 5: Limitations of using K-Means for anomaly detection



Assumes spherical, equally-sized clusters – K-Means assumes clusters are round and evenly distributed. Real fraud patterns are often irregular or scattered, so this assumption doesn't always hold.

Requires choosing K in advance – The number of clusters (K=10 here) is chosen arbitrarily. A wrong K value can distort distance-based anomaly scoring.

Sensitive to feature scaling – Since K-Means relies on Euclidean distance, unscaled features (like Amount) can dominate the distance calculation and mislead the clustering.

Not designed for imbalanced data – K-Means clusters based on the majority pattern (normal transactions), so rare fraud cases may get absorbed into a normal-looking cluster instead of standing out.

Distance ≠ fraud indicator – Some frauds are designed to closely resemble normal transactions (to avoid detection), so they may end up with small distances and go undetected — meaning K-Means will inherently miss "disguised" fraud.

No probabilistic confidence – K-Means gives a hard distance value, not a probability of fraud, making it harder to justify or explain individual predictions (a concern for financial audits).

Sensitive to outliers during training – A few extreme values can shift cluster centers, distorting the boundaries used for anomaly detection.
