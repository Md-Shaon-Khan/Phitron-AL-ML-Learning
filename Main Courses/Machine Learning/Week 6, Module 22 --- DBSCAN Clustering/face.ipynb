```python
import pandas as np
import numpy as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.datasets import make_circles
from sklearn.cluster import DBSCAN
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
```

```python
X,y = make_circles(n_samples=5000, noise=.01, random_state=42)
```

```python
sns.scatterplot(x=X[:,0], y=X[:,1])
```

```python
scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)
```

```python
dbscan = DBSCAN(eps=0.1, min_samples=5)

dbscan.fit(X_scaled)
label_dbscan = dbscan.labels_
```

```python
sns.scatterplot(x=X_scaled[:,0], y=X_scaled[:,1], hue=label_dbscan, palette='viridis')
plt.show()
```

```python
kmeans = KMeans(n_clusters=2, init='k-means++', max_iter=500, random_state=42)
kmeans.fit(X_scaled)
label_kmeans= kmeans.labels_
```

```python
sns.scatterplot(x=X_scaled[:,0], y=X_scaled[:,1], hue=label_kmeans, palette='viridis')
plt.show()
```
