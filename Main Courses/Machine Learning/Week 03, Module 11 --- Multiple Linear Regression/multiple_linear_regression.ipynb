```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
sns.set_theme()
```

```python
X = np.array([
    [1, 4, 1.5, 3.8],
    [3, 6, 1.0, 3.1],
    [4, 7, 0.8, 3.9],
    [6, 8, 0.4, 3.4],
    [7, 9, 0.2, 2.8]
])

y = np.array([45, 60, 72, 85, 95])
```

```python
sns.scatterplot(x = X[:, 0:1].ravel(), y = y)
plt.xlabel('Years of Experience')
plt.ylabel('Salary')
plt.show()
```

```python
sns.scatterplot(x = X[:, 1:2].ravel(), y = y)
plt.xlabel('Problem Solving Skills')
plt.ylabel('Salary')
plt.show()
```

```python
sns.scatterplot(x = X[:, 2:3].ravel(), y = y)
plt.xlabel('Pause Time during Interview')
plt.ylabel('Salary')
plt.show()
```

```python
sns.scatterplot(x = X[:, 3:].ravel(), y = y)
plt.xlabel('CGPA')
plt.ylabel('Salary')
plt.show()
```
