```python
pclass = df['Pclass'].value_counts()

sns.countplot(data = df, x = 'Pclass')
```

```python
plt.hist(df['Age'],bins=50, edgecolor='black')

plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.grid(alpha = 0.3)
plt.show()
```

```python
sns.kdeplot(df['Age'])
plt.title('KDE Plot of Age')
plt.xlabel('Age')
plt.ylabel('Density')
plt.show()
```
