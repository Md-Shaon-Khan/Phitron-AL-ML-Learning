```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("F:\Jupyter_Projects\Phitron AL ML\Dataset\house price 3 columns.csv")
df.sample(5)
```

```python
house_size = np.array(df['area'])
house_price = np.array(df['price'])

house_size
```

```python
house_price
```

```python
sns.scatterplot(x = house_size, y = house_price)
```

```python
def make_prediction(area, price, w, b):
    n = area.shape[0]
    y_prediction = np.zeros(n)
    for i in range(n):
        y_prediction[i] = w * area[i] + b
    return y_prediction
```

```python
pred = make_prediction(house_size, house_price, 100, 10000)

sns.scatterplot(x = house_size, y = house_price)
plt.plot(house_size, pred, color = 'red')
```

```python
pred = make_prediction(house_size, house_price, 148, 20000)

sns.scatterplot(x = house_size, y = house_price)
plt.plot(house_size, pred, color = 'red')
```

```python
from sklearn.linear_model import LinearRegression
```

```python
lr = LinearRegression()
lr.fit(house_size.reshape(-1, 1), house_price)
w, b = lr.coef_[0], lr.intercept_
print(w, b)
```

```python
pred = make_prediction(house_size, house_price, w, b)

sns.scatterplot(x = house_size, y = house_price)
plt.plot(house_size, pred, color = 'red')
```
