```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('enrollment_data.csv')
df
```

```python
plt.xlabel('Year')
plt.ylabel('Count of enrollment')
plt.title('Student enrollment over years')

plt.plot(df['Year'], df['Programming'])
plt.plot(df['Year'], df['Digital Marketing'])
```

```python
plt.xlabel('Year')
plt.ylabel('Count of enrollment')
plt.title('Student enrollment over years')

plt.plot(df['Year'], df['Programming'], label = 'Programming')
plt.plot(df['Year'], df['Digital Marketing'], label = 'DM')
plt.legend(loc = 'lower right')
```

```python
plt.xlabel('Year')
plt.ylabel('Count of enrollment')
plt.title('Student enrollment over years')

plt.plot(df['Year'], df['Programming'], label = 'Programming', color = 'black', linewidth = 5)
plt.plot(df['Year'], df['Digital Marketing'], label = 'DM', color = '#32a852')
plt.legend(loc = 'lower right')
```
