```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("E:\Down\Bank Customer Churn Prediction.csv")
df.isnull().sum()
```

```python
def add_nulls(data, column, percentage):
    # মোট ডেটার কত শতাংশ Null হবে তা নির্ধারণ করা
    num_nulls = int(len(data) * percentage)
    # র্যান্ডম ইনডেক্স নির্বাচন করা
    random_indices = np.random.choice(data.index, num_nulls, replace=False)
    # নির্বাচিত ইনডেক্সগুলোতে NaN বসানো
    data.loc[random_indices, column] = np.nan
    return data

# ৩. 'age' এবং 'balance' কলামে ১০% Null ভ্যালু যোগ করা
df = add_nulls(df, 'age', 0.0328)
df = add_nulls(df, 'balance', 0.0873)

# ৪. নতুন ফাইল হিসেবে সেভ করা
df.to_csv('bank_with_missing.csv', index=False)

print("সফলভাবে 'age' এবং 'balance' কলামে ১০% মিসিং ভ্যালু যোগ করা হয়েছে।")
print("নতুন ফাইলটি 'bank_with_missing.csv' নামে সেভ করা হয়েছে।")

# ৫. ভেরিফিকেশন (Null চেক)
print("\n--- নতুন ডেটাসেটে Null ভ্যালুর সংখ্যা ---")
print(df.isnull().sum())
```
