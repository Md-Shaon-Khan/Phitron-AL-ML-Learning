```python
# ==========================================
# SIMPLE NEURAL NETWORK FROM SCRATCH
# ==========================================

# ------------------------------------------
# 1. DATA
# ------------------------------------------

X = [
    [2.0, 3.0],
    [5.0, 6.0],
    [7.0, 8.0]
]

y = [
    3.0,
    4.0,
    5.0
]


# ------------------------------------------
# 2. INITIAL WEIGHTS AND BIASES
# ------------------------------------------

# First layer
w11_1 = 0.1
w21_1 = 0.1

w12_1 = 0.1
w22_1 = 0.1

b1_1 = 0.0
b2_1 = 0.0


# Second layer
w11_2 = 0.1
w21_2 = 0.1

b_2 = 0.0


# Learning rate
lr = 0.001


# ------------------------------------------
# 3. TRAINING
# ------------------------------------------

epochs = 50

for epoch in range(epochs):

    total_loss = 0


    # Take one sample at a time
    for i in range(len(X)):

        # ----------------------------------
        # INPUT
        # ----------------------------------

        x1 = X[i][0]
        x2 = X[i][1]

        actual_y = y[i]


        # ==================================
        # FORWARD PROPAGATION
        # ==================================

        # Hidden neuron 1
        #
        # a1 = w11^1*x1 + w21^1*x2 + b1^1

        a1 = (
            w11_1 * x1
            + w21_1 * x2
            + b1_1
        )


        # Hidden neuron 2
        #
        # a2 = w12^1*x1 + w22^1*x2 + b2^1

        a2 = (
            w12_1 * x1
            + w22_1 * x2
            + b2_1
        )


        # Output
        #
        # y_hat = w11^2*a1 + w21^2*a2 + b^2

        y_hat = (
            w11_2 * a1
            + w21_2 * a2
            + b_2
        )


        # ==================================
        # LOSS
        # ==================================

        # L = (y - y_hat)^2

        loss = (actual_y - y_hat) ** 2

        total_loss += loss


        # ==================================
        # BACKPROPAGATION
        # ==================================

        # ----------------------------------
        # Step 1
        # L -> y_hat
        # ----------------------------------

        # dL/dy_hat = -2(y - y_hat)

        dL_dyhat = -2 * (actual_y - y_hat)


        # ----------------------------------
        # Step 2
        # y_hat -> output weights
        # ----------------------------------

        # dL/dw11^2
        dL_dw11_2 = dL_dyhat * a1

        # dL/dw21^2
        dL_dw21_2 = dL_dyhat * a2

        # dL/db^2
        dL_db_2 = dL_dyhat


        # ----------------------------------
        # Step 3
        # y_hat -> a1 and a2
        # ----------------------------------

        # dL/da1
        dL_da1 = dL_dyhat * w11_2

        # dL/da2
        dL_da2 = dL_dyhat * w21_2


        # ----------------------------------
        # Step 4
        # a1 -> first layer weights
        # ----------------------------------

        # dL/dw11^1
        dL_dw11_1 = dL_da1 * x1

        # dL/dw21^1
        dL_dw21_1 = dL_da1 * x2

        # dL/db1^1
        dL_db1_1 = dL_da1


        # ----------------------------------
        # Step 5
        # a2 -> first layer weights
        # ----------------------------------

        # dL/dw12^1
        dL_dw12_1 = dL_da2 * x1

        # dL/dw22^1
        dL_dw22_1 = dL_da2 * x2

        # dL/db2^1
        dL_db2_1 = dL_da2


        # ==================================
        # WEIGHT UPDATE
        # ==================================

        # w_new = w_old - learning_rate * gradient


        # ----------------------------------
        # Second layer
        # ----------------------------------

        w11_2 = w11_2 - lr * dL_dw11_2

        w21_2 = w21_2 - lr * dL_dw21_2

        b_2 = b_2 - lr * dL_db_2


        # ----------------------------------
        # First layer
        # ----------------------------------

        w11_1 = w11_1 - lr * dL_dw11_1

        w21_1 = w21_1 - lr * dL_dw21_1

        w12_1 = w12_1 - lr * dL_dw12_1

        w22_1 = w22_1 - lr * dL_dw22_1


        # ----------------------------------
        # First layer biases
        # ----------------------------------

        b1_1 = b1_1 - lr * dL_db1_1

        b2_1 = b2_1 - lr * dL_db2_1


    # ======================================
    # EPOCH LOSS
    # ======================================

    average_loss = total_loss / len(X)

    print(
        f"Epoch {epoch + 1} "
        f"Loss = {average_loss:.4f}"
    )
```

```python
print("\nFinal Parameters:")

print("w11^1 =", w11_1)
print("w21^1 =", w21_1)

print("w12^1 =", w12_1)
print("w22^1 =", w22_1)

print("b1^1  =", b1_1)
print("b2^1  =", b2_1)

print("w11^2 =", w11_2)
print("w21^2 =", w21_2)

print("b^2   =", b_2)
```
